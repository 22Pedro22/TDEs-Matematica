## Teste 1 - Fundamentos Matemáticos

### 1. Contextualização

Uma empresa de comércio eletrônico acompanha seus custos mensais de transporte. A empresa deseja analisar como esses custos estão variando ao longo do tempo e utilizar os dados históricos para realizar uma previsão dos custos dos próximos meses.

### 2. Objetivo

Utilizar conceitos de funções para analisar os dados, representar graficamente o comportamento dos custos e realizar uma previsão para os próximos meses.

### 4. Atividades - Etapa 1: Carregar os dados

```python
# TESTE 1: FUNDAMENTOS MATEMÁTICOS - ANÁLISE DE CUSTOS DE TRANSPORTE

# Conexão com o drive
from google.colab import drive
drive.mount('/content/drive')

# Defina o caminho exato onde o arquivo 'base_custos_transporte.csv' está salvo no seu Drive.
# Se o arquivo estiver na pasta raiz do seu Drive, mantenha o caminho abaixo:
caminho_csv = '/content/drive/MyDrive/base_custos_transporte.csv'

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Etapa 1 (carregar dados)

print("="*60)
print("ETAPA 1: CARREGAR E APRESENTAR A TABELA DE DADOS")
print("="*60)

try:
    df = pd.read_csv(caminho_csv)
    print("✓ Arquivo CSV carregado com sucesso do Google Drive!\n")
except Exception as e:
    print(f"⚠ Não foi possível abrir o arquivo no caminho do Drive especificado.")
    print("Usando dados de contingência incorporados ao código:\n")
    data = {
        'mes': list(range(1, 13)),
        'custo_transporte': [44500, 45200, 48300, 48600, 51300, 53300, 53800, 57000, 57900, 61000, 61300, 64000]
    }
    df = pd.DataFrame(data)

# Exibe a tabela formatada
print(df.to_string(index=False))

# Etapa 2 (identificar variárveis)

print("\n" + "="*60)
print("ETAPA 2: IDENTIFICAÇÃO DAS VARIÁVEIS")
print("="*60)
print("a) Variável independente (x): Mês ('mes')")
print("b) Variável dependente (y): Custo de Transporte ('custo_transporte')")
print("c) Representação no contexto:")
print("   - 'mes' (x): Representa o tempo decorrido ao longo do período de 12 meses.")
print("   - 'custo_transporte' (y): Representa o custo total em Reais (R$) de frete no mês.")

# Etapa 4: (encontrar a função afim f(x) = ax + b)

x = df['mes'].values
y = df['custo_transporte'].values

# Ajuste linear pelo Método dos Mínimos Quadrados
a, b = np.polyfit(x, y, 1)

print("\n" + "="*60)
print("ETAPA 4: ENCONTRAR A FUNÇÃO AFIM f(x) = ax + b")
print("="*60)
print(f"• Coeficiente 'a' (inclinação): {a:.2f}")
print(f"• Coeficiente 'b' (intercepto):  {b:.2f}")
print(f"• Função encontrada: f(x) = {a:.2f}x + {b:.2f}")

# Etapa 5 (interpretar a função)

print("\n" + "="*60)
print("ETAPA 5: INTERPRETAÇÃO DOS COEFICIENTES")
print("="*60)
print(f"a) Coeficiente 'a' ({a:.2f}): Representa a taxa de variação média mensal do custo de transporte (R$ {a:.2f}/mês).")
print(f"b) Coeficiente 'b' ({b:.2f}): Representa o custo fixo inicial estimado no ponto zero (R$ {b:.2f}).")
print("c) Como 'a' é positivo (a > 0), o valor indica CRESCIMENTO constante dos custos.")

# Etapas 3 e 6: Construir gráficos com dados reais e a reta de tendência

print("\n" + "="*60)
print("ETAPAS 3 & 6: GRÁFICO DE DISPERSÃO E RETA DE TENDÊNCIA")
print("="*60)

plt.figure(figsize=(10, 6))

# Plota os dados reais
plt.scatter(df['mes'], df['custo_transporte'], color='blue', label='Dados Reais', s=70, zorder=5)

# Plota a reta f(x) = ax + b
x_reta = np.linspace(1, 12, 100)
y_reta = a * x_reta + b
plt.plot(x_reta, y_reta, color='red', linestyle='--', linewidth=2, label=f'Tendência: f(x) = {a:.2f}x + {b:.2f}')

plt.title('Evolução do Custo Mensal de Transporte e Tendência Linear', fontsize=14, fontweight='bold')
plt.xlabel('Mês (x)', fontsize=12)
plt.ylabel('Custo de Transporte (R$)', fontsize=12)
plt.xticks(range(1, 13))
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()

print("\nDescrição do comportamento observado:")
print("Observa-se uma tendência de crescimento linear positivo e constante nos custos ao longo dos 12 meses.")

# Etapa 7: fazer previsões (meses 13, 14 e 15)

print("\n" + "="*60)
print("ETAPA 7: PREVISÕES PARA OS PRÓXIMOS 3 MESES")
print("="*60)

meses_futuros = [13, 14, 15]
previsoes = {}

for m in meses_futuros:
    custo_est = a * m + b
    previsoes[m] = custo_est
    print(f"• Mês {m}: f({m}) = {a:.2f} * {m} + {b:.2f} = R$ {custo_est:,.2f}")

# Etapa 8: (analisar o resultado)

r2 = np.corrcoef(x, y)[0, 1]**2

print("\n" + "="*60)
print("ETAPA 8: ANÁLISE DOS RESULTADOS")
print("="*60)
print(f"a) A função representa muito bem a tendência dos dados (R² = {r2:.4f}, ou seja, ~99,05% de precisão explicativa).")
print("b) Os custos estão AUMENTANDO ao longo do tempo.")
print(f"c) O custo aumenta aproximadamente R$ {a:.2f} a cada mês.")

# Etapa 9 (conclusão final)

print("\n" + "="*60)
print("ETAPA 9: CONCLUSÃO FINAL")
print("="*60)
print("Com base nos dados analisados, a empresa pode esperar um aumento contínuo nos seus custos de transporte.")
print(f"Para os próximos três meses (13, 14 e 15), os custos estimados são de R$ {previsoes[13]:,.2f}, R$ {previsoes[14]:,.2f} e R$ {previsoes[15]:,.2f}, respectivamente.")
print("Recomenda-se um planejamento orçamentário preventivo e avaliação de estratégias de mitigação do frete.")
```
